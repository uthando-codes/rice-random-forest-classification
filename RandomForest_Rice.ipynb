```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    ConfusionMatrixDisplay
)

print("All libraries imported successfully.")
```

```python
df = pd.read_excel("Rice_Cammeo_Osmancik.xlsx")

print(f"Dataset shape: {df.shape}")
print(f"\nClass distribution:")
print(df["Class"].value_counts())
df.head()
```

```python
# check and explore the data
print("Data types:")
print(df.dtypes)
print("\nMissing values:")
print(df.isnull().sum())
print("\nSummary statistics:")
df.describe()
```

```python
# Encode 
le = LabelEncoder()
df["Class_encoded"] = le.fit_transform(df["Class"])

print("Label mapping:")
for label, code in zip(le.classes_, le.transform(le.classes_)):
    print(f"  {label} → {code}")
```

```python
# Train and test split 
feature_cols = [c for c in df.columns if c not in ("Class", "Class_encoded")]
X = df[feature_cols].values
y = df["Class_encoded"].values

# 80 % train, 20 % test 
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print(f"Training samples : {X_train.shape[0]}")
print(f"Test samples     : {X_test.shape[0]}")
print(f"\nFeatures used: {feature_cols}")
```

```python
#Train random classifier
rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=None,
    random_state=42,
    n_jobs=-1          # use all CPU cores
)
rf.fit(X_train, y_train)
print("Model trained successfully.")
```

```python
# evaluate confusion matrix and metrics 
# Prediction
y_pred = rf.predict(X_test)

# 1. Confusion Matrix 
cm = confusion_matrix(y_test, y_pred)
print("=" * 50)
print("1. CONFUSION MATRIX")
print("=" * 50)
print(f"\n{'':>20} Predicted")
print(f"{'':>12} {'Cammeo':>10} {'Osmancik':>10}")
print(f"Actual Cammeo   {cm[0,0]:>10}  {cm[0,1]:>10}")
print(f"       Osmancik {cm[1,0]:>10}  {cm[1,1]:>10}")
print()
print(f"  TN (Cammeo correct)    : {cm[0,0]}")
print(f"  FP (Cammeo → Osmancik) : {cm[0,1]}")
print(f"  FN (Osmancik → Cammeo) : {cm[1,0]}")
print(f"  TP (Osmancik correct)  : {cm[1,1]}")

#2. Accuracy 
accuracy = accuracy_score(y_test, y_pred)
print("\n" + "=" * 50)
print("2. ACCURACY")
print("=" * 50)
print(f"   Accuracy = {accuracy:.4f}  ({accuracy*100:.2f}%)")
print("   Formula  : (TP + TN) / (TP + TN + FP + FN)")

# 3. Precision 
precision = precision_score(y_test, y_pred)
print("\n" + "=" * 50)
print("3. PRECISION")
print("=" * 50)
print(f"   Precision = {precision:.4f}  ({precision*100:.2f}%)")
print("   Formula   : TP / (TP + FP)")

#  4. Sensitivity (Recall) 
sensitivity = recall_score(y_test, y_pred)
print("\n" + "=" * 50)
print("4. SENSITIVITY (RECALL)")
print("=" * 50)
print(f"   Sensitivity = {sensitivity:.4f}  ({sensitivity*100:.2f}%)")
print("   Formula     : TP / (TP + FN)")

#  5. F1 Score 
f1 = f1_score(y_test, y_pred)
print("\n" + "=" * 50)
print("5. F1 SCORE")
print("=" * 50)
print(f"   F1 Score = {f1:.4f}  ({f1*100:.2f}%)")
print("   Formula  : 2 × (Precision × Sensitivity) / (Precision + Sensitivity)")
```

```python
#  summary table 
print("\n" + "=" * 50)
print("       RESULTS SUMMARY")
print("=" * 50)
metrics = {
    "Accuracy"   : accuracy_score(y_test, y_pred),
    "Precision"  : precision_score(y_test, y_pred),
    "Sensitivity": recall_score(y_test, y_pred),
    "F1 Score"   : f1_score(y_test, y_pred),
}
for name, val in metrics.items():
    print(f"  {name:<14}: {val:.4f}  ({val*100:.2f}%)")
print("=" * 50)
```
